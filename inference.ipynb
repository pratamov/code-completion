```python
import json
from inference import load_model, make_inference
from tokenizer import tokenize
from detokenizer import detokenize

model, vocab_data, metadata = load_model()

with open("vocab_data.json", "w", encoding="utf-8") as f:
    json.dump(vocab_data, f, ensure_ascii=False, indent=4)


NUM_TOKEN_INPUT = 10
NUM_TOKEN_PREDICTION = 5
TEMPERATURE = 0.9
with open("snippets/DemoDashboard2.SRC", "r", encoding="utf-8") as f:
    text = f.read()
    namespaces, properties = tokenize(text)
    namespaces = namespaces.split(" ")
    properties = properties.split(" ")
    assert len(namespaces) == len(properties), "Error tokenizer: namespaces and properties has different length"
    token_size = len(namespaces)
    for i in range(0, len(namespaces)//NUM_TOKEN_INPUT, NUM_TOKEN_INPUT):
        print("---------------------")
        seed_tokens = namespaces[i:min(token_size, i+NUM_TOKEN_INPUT)]
        seed_properties = [properties[i:min(token_size, i+NUM_TOKEN_INPUT)]]
        print(namespaces[i:min(token_size, i+NUM_TOKEN_INPUT)])
        print(properties[i:min(token_size, i+NUM_TOKEN_INPUT)])
        generated_words, generated_properties = make_inference(model=model, vocab_data=vocab_data, 
                       seed_tokens=seed_tokens, seed_properties=seed_properties, 
                       num_generate=NUM_TOKEN_PREDICTION, temperature=TEMPERATURE)
        
        print(generated_words)
        print(generated_properties)
        
        if i > 3: break
        
```
